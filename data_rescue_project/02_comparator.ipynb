```python
import pandas as pd

# assume df_drp and df_comp already loaded

df_comp_sample = df_comp.sample(
    n=len(df_drp),
    random_state=42
)

df_balanced = pd.concat([df_drp, df_comp_sample], ignore_index=True)

print(df_balanced["label"].value_counts())
```

```python
df_balanced.groupby("label")["n_banned_terms_total"].mean()
```

```python
df_drp.groupby("label")["has_flagged_word_total"].mean()
```

```python
drp_rate = df_drp["has_flagged_word_total"].mean()

comp_with = df_comp[df_comp["has_flagged_word_total"] == 1]
comp_without = df_comp[df_comp["has_flagged_word_total"] == 0]

n_without = len(comp_without)

target_n = len(df_drp)
n_with = min(len(comp_with), target_n - n_without)

comp_balanced = pd.concat([
    comp_with.sample(n=n_with, random_state=42),
    comp_without
], ignore_index=True)

print(comp_balanced["has_flagged_word_total"].value_counts())
print(comp_balanced.shape)
```

```python
comp_with = df_comp[df_comp["has_flagged_word_total"] == 1]
comp_without = df_comp[df_comp["has_flagged_word_total"] == 0]

n_without = len(comp_without)
n_with = n_without

comp_balanced = pd.concat([
    comp_with.sample(n=n_with, random_state=42),
    comp_without.sample(n=n_without, random_state=42)
], ignore_index=True)

print(comp_balanced["has_flagged_word_total"].mean())
print(comp_balanced.shape)
```
